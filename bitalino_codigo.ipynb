```python
#Importamos librerías

import bitalino

import csv

from IPython.display import HTML, display, Javascript
```

```python
#Configuración del Bitalino

PUERTO = "COM7"# Puerto Bluetooth correspondiente al Bitalino

CANAL = 0# A1 = 0, A2 = 1, A3 = 2, etc...

PERIODO = 0.01# (1,0.1,0.01) Cantidad de segundos en que se demora en realizar las mediciones de forma consecutiva desde el inicio del experimento

N = 10# Número de mediciones realizadas en bloques, cuando se capturen las {N} observaciones serán enviadas al dispotivo receptor para ser
#almacenadas en el .CSV y visualizadas en tiempo real.

#Si {N} = 1 entonces los datos se envian actualizan el CSV y visualizador en tiempo real, si es mayor a 0 entonces se espera a juntar {N}
#observaciones para enviar los datos, actualizar el CSV y el visualizador.

DURACION = 10# Cantidad de segundos en que se demora en realizarse el experimento
```

```python
with open("visualizador.html", "r", encoding="utf-8") as archivo:
    Visualizador_HTML = archivo.read()
```

```python
device = bitalino.BITalino(PUERTO)#abre la conexión con el BITalino usando el puerto indicado {PUERTO}
```

```python
#ECG

#Datos Iniciales del Visualizador

titulo_grafico = "Serie temporal del ECG"#Nombre del gráfico

etiqueta_eje_y = "ECG"#Nombre del Eje y del gráfico

archivo_csv = "ECG_100s.csv"#Nombre del Archivo .csv dónde se almacenaran los datos


#Inyectamos el titulo del gráfico, la etiqueta del eje y, el tiempo de duración parámetros al Visualizador del HTML

Visulizador_HTML_Inicial = (
    Visualizador_HTML
    .replace("__DURACION__", str(DURACION))
    .replace("__TITULO__", titulo_grafico)
    .replace("__ETIQUETA_EJE_Y__", etiqueta_eje_y)
)


# Inicializamos el Visualizador

display(HTML(Visulizador_HTML_Inicial))

actualizador = display(Javascript("void 0;"), display_id=True)

n_medicion = 0#Observación número {n}



try:


    with open(archivo_csv, "w", newline="", encoding="utf-8") as archivo:#Crea EL CSV
        escritor_csv = csv.writer(archivo)#Crea un objeto para escribir filas con formato CSV
        escritor_csv.writerow(["Tiempo (s)", "Magnitud"])#Crea las columnas del CSV

        device.start(1 / PERIODO, [CANAL])#Bitalino empezará a medir el canal ECG cada {PERIODO} segundos

        N_mediciones_totales = int(DURACION / PERIODO)#Número de mediciones totales que realizará el Bitalino

        while n_medicion < N_mediciones_totales:
            cantidad = min(N, N_mediciones_totales - n_medicion)#Número de mediciones que se realizarán para completar un bloque
            mediciones_bloque = []#Mediciones de un bloque correspondiente

            for muestra in device.read(cantidad):
                n_medicion += 1

                tiempo = n_medicion * PERIODO#Tiempo en que en Bitalino realizó la medición
                valor = int(muestra[5])#Magnitud de la medición 

                mediciones_bloque.append([tiempo, valor])#Registramos los datos dentro del bloque correspondiente
                escritor_csv.writerow([tiempo, valor])#Registramos los datos en el .csv

            archivo.flush()
        
            actualizador.update(Javascript(f"window.ecgInyectar({mediciones_bloque});"))

finally:
    device.close()
```
